```python
# Tip Calculator Practice

bill = float(input("Enter your bill amount: $"))

print("Your bill is: $", bill)
```

```python
# Ask the user what percentage they want to tip
tip_percent = float(input("What percentage would you like to tip? "))

# Calculate the tip amount
tip = bill * (tip_percent / 100)

# Display the tip
print("Your tip is: $", round(tip, 2))
```

```python
# Calculate the total bill including tip
total = bill + tip

# Display the final amount
print("Original bill: $", round(bill, 2))
print("Tip amount: $", round(tip, 2))
print("Total including tip: $", round(total, 2))
```

```python
# Ask how many people are splitting the bill
people = int(input("How many people are splitting the bill? "))

# Calculate how much each person owes
per_person = total / people

# Display the amount per person
print("Each person owes: $", round(per_person, 2))
```
